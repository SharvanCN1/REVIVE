```python
import numpy as np
```

```python
class blur():

    def __init__(self, img):
        self.img_matrix = img
        self.img_h = img.shape[0]
        self.img_w = img.shape[1]

    def gaussian_blur(self, std_dev = 1):
        self.std_dev = std_dev
        self.f_2D = lambda x, y: 1 / (2 * np.pi * self.std_dev**2) * np.e**(-(x**2 + y**2) / (2 * self.std_dev**2))
        self.f_1D = lambda x: 1 / np.sqrt(2 * np.pi * self.std_dev**2) * np.e**( -x**2 / (2 * self.std_dev**2))
```
